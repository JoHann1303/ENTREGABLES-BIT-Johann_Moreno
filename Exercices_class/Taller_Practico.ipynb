```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/shaadyalii/Dirty-Cafe-Sales-Dataset/main/dirty_cafe_sales.csv"

df = pd.read_csv(url)

df.head()
```

```python
for col in df.columns:
    error_count = df[col].isin(["ERROR", "UNKNOWN"]).sum()
    if error_count > 0:
         print(f"{col}, {error_count}" )

```

```python
df.replace(["ERROR", "UNKNOWN"], np.nan, inplace=True)
```

```python
columnas_numericas =["Quantity","price per unit","tortal spent"]

for col in columnas_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df[columnas_numericas].dtypes
```
