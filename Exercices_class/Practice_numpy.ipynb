```python
import numpy as np

ventas = np.array([120, 150, 90, 200, 180])
#Actividad
#Completa el código para calcular:
#Las ventas con un descuento equivalente al 5 % del valor original.
ventas_ajustadas = ventas-(ventas*0.05)
ventas_ajustadas

#Las ventas incrementadas en 25 unidades.
ventas_ajustadas_incrementadas = ventas + 25
ventas_ajustadas_incrementadas

#El doble de cada valor vendido.
ventas_doble = ventas * 2
print(f"Las ventas dobles son: {ventas_doble}",
      "las ventas ajustadas son: ",ventas_ajustadas,
      "las ventas ajustadas incrementadas son: ",ventas_ajustadas_incrementadas)
```

```python
ventas_sucursales = np.array([
    [120, 150, 180],  # Sucursal A
    [90,  200, 170],  # Sucursal B
    [160, 140, 210]   # Sucursal C
])


# Escribe tu código debajo de cada comentario.
# 1. Tercera sucursal
print("tercera sucursal:", ventas_sucursales[2])

# 2. Primer día de todas las sucursales
print("Primer día de todas las sucursales:", ventas_sucursales[:, 0])
# 3. Segundo día de la sucursal B
print("Segundo día de la sucursal B:", ventas_sucursales[1, 1])
```

```python
import numpy as np
# 1. Ventas inferiores a 100
ventas_inferiores_100 = ventas_sucursales[ventas_sucursales < 100]
print("Ventas inferiores a 100:", ventas_inferiores_100)

# 2. Ventas iguales o superiores a 180
ventas_iguales_superiores_180 = ventas_sucursales[ventas_sucursales >= 180]
print("Ventas iguales o superiores a 180:", ventas_iguales_superiores_180)

# 3. Porcentaje de días con ventas inferiores a 100
total_días = np.mean(ventas_sucursales < 100)*100
print("porcentaje Total de días:", total_días )
```
